## 1. Ôn nhanh kiến thức

- `== != < <= > >=` so sánh; `and/or/not` kết hợp điều kiện.

- Khối code dùng thụt lề nhất quán, thường 4 dấu cách.

## Bài A · Điền nhánh xếp loại (cơ bản)
Điểm 0–10: >=8 là `Tot`, >=5 là `Dat`, còn lại `Can co gang`. Dùng if/elif/else, không viết trực tiếp đáp án cho điểm mẫu.


```python
score = 8
classification = None
# TODO: dung if/elif/else de gan classification
if 0 < score and score < 10:
    if score >= 8:
        classification = "Tot"
    elif score >= 5:
        classification = "Dat"
    else:
        classification = "Can co gang"
else:
    print(classification)
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if classification is None:
    print("Chua hoan thanh Bai A.")
else:
    assert classification == "Tot"
    print("Dat mau diem 8. Tu thu cac moc 0, 5, 7.9, 10.")
```

## Bài B · Tự viết vòng lặp (cơ bản)
Tính tổng các số chẵn từ 1 đến n=10 bằng for. Không dùng công thức hoặc gán trực tiếp 30. Sau đó tự thử n=0 và n=1 → tổng 0.


```python
n = 10
even_total = 0
# TODO: khoi tao tong va viet vong lap
for i in range(1,n+1):
    if i%2==0:
        even_total += i
        print(f"{i=}, {even_total}")
    
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if even_total is None:
    print("Chua hoan thanh Bai B.")
else:
    assert even_total == 30
    print("Dat du lieu mau; tu thu n=0 va n=1.")
```

## Bài C · while và điều kiện dừng (cơ bản)
Dùng while in đếm ngược 5,4,3,2,1; đếm số lần lặp vào turns. Sau vòng lặp, counter=0 và turns=5.


```python
counter = 5
turns = 0
# TODO: while, in counter, giam counter, tang turns
while counter > 0:
    print(counter)
    counter -= 1
    turns += 1
print(f"counter = {counter}, turns = {turns}")
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if turns is None:
    print("Chua hoan thanh Bai C.")
else:
    assert counter == 0 and turns == 5
    print("Dat Bai C.")
```

## Bài D · Thử thách
Viết trò đoán số 1–20, giới hạn 5 lượt, thông báo “cao hơn/thấp hơn”, dừng ngay khi đoán đúng. Bản kiểm thử dùng secret=7 để biết kết quả; khi chơi mới dùng random. Giả định đầu vào là số nguyên hợp lệ, chưa cần try/except. Ghi rõ vị trí dùng break và điều kiện dừng.


```python
# TODO: viet moi tro doan so; secret = 7 khi tu kiem tra
secret = 7

for i in range(5):
    a = int(input("Nhập số mà bạn đoán: "))
    if a == secret :
        print("đoán đúng")
        break
    elif a < secret:
        print("Thấp hơn")
    else:
        print("Cao hơn")
    
    
```
