## Ví dụ · if / elif / else
Mỗi lần chạy, chương trình chọn nhánh đúng đầu tiên trong chuỗi dưới đây. Quan sát dấu `:` và thụt lề bốn dấu cách.

````python
temperature = 28
if temperature >= 35:
    print("Nóng")
elif temperature >= 25:
    print("Ấm")
else:
    print("Mát")
````

## Bài 3 · Tổng từ 1 đến n 
Cho số nguyên `n >= 0`. Dùng `for` và `range` tính tổng từ 1 đến n, lưu vào `total_b3`.

Không dùng `sum()` hoặc công thức tính nhanh. Khởi tạo tổng **trước vòng lặp**. Có thể viết `total_b3 += i`, tương đương `total_b3 = total_b3 + i` trong bài này.

Với `n = 5`, kết quả là `15`.

````python
n_b3 = 5
total_b3 = 0

# Viết vòng lặp tại đây.

print(total_b3)
````

## Ví dụ · break và continue
Chạy từng ô và so sánh. `break` thoát vòng lặp gần nhất. `continue` bỏ phần còn lại của lượt hiện tại.

````python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
print("Đã ra khỏi vòng lặp")
````

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````
